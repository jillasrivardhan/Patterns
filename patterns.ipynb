### star - patterns

```python
n = 4
for i in range(1,n+1):
  for j in range(1,n):
    print('*',end="")
  print("\n")
```

```python
n = 4
for i in range(1,n):
  for j in range(1,n+1):
    print('*',end="")
  print("\n")
```

```python
n = 5
for i in range(1,n+1):
    print("*" * i)
```

```python
n = 5
for i in range(1,n+1):
    print(" "*(n-i)+"*"*i)
```

```python
n = 5
for i in range(5,0,-1):
    print("*"*i)
```

```python
n = 5
for i in range(5,0,-1):
    print(" "*(n-i)+"*"*i)
```
